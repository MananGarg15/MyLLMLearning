```python
llama = OpenAI(base_url="http://localhost:11434/v1/",api_key=os.getenv('GOOGLE_API_KEY'))
```

```python
links_system_prompt = """
You will be provided with a set of links from a website. You have to correctly identify the links most relevant to be included in the brochure
 of the company and provide them in as a JSON object, specifying what they are about. Do not include terms of service, email links.
Here is an example - 
{
    'links' : [ {'type':'about page','url':'https://fullurl/goes/here/about'},
                {'type':'home page','url':'https://anotherurl/goes/here/home'}
    ]
}

    """
links = get_all_links('https://edwarddonner.com/')

links_user_prompt = f"""
Here is the list of links in https://edwarddonner.com/- 

{links}    
    """
```

```python
response = Llms.llama.chat.completions.create(model='llama3.2',messages=[{'role':'system','content':links_system_prompt},
                                                                                        {'role':'user','content':links_user_prompt}],
                                                                                        response_format={'type':'json_object'}) 
```

```python


result = response.choices[0].message.content
```
