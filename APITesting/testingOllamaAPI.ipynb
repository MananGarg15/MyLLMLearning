```python
OLLAMA_BASE_URL = "http://localhost:11434/v1"
messages =[{'role':'user','content':'Hi. This is my first time talking to you.'},
{'role':'user','content':'Do you see this line as a separate prompt from previous one? or is it one single prompt for you?'},
{'role':'user','content':'What prompt have you received from me till now? Tell me exactly.'}]
```

```python
llama = OpenAI(base_url=OLLAMA_BASE_URL,api_key='anything')
```

```python
response = llama.chat.completions.create(model='llama3.2',messages=messages)
print(response.choices[0].message.content)
```

```python
response = llama.chat.completions.create(model='llama3.2',messages=[{'role':'user','content':'what was your last prompt from me?'}])
print(response.choices[0].message.content)
```
